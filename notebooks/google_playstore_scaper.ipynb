```python
from google_play_scraper import reviews, Sort
import pandas as pd
import time
from datetime import datetime
```

```python
# Target aplikasi
package_name = 'com.miHoYo.GenshinImpact'

# Tentukan rentang tanggal review yang ingin diambil
start_date = datetime(2025, 2, 12)  # Tanggal mulai

end_date = datetime(2025, 3, 25)        # Gunakan Line ini jika tanggal yang diinginkan adalah custom
# end_date = datetime.today()         # Gunakan Line ini jika tanggal yang diinginkan adalah hari ini

version = '5.4'  # Versi dari aplikasi

# Penyimpanan review
all_reviews = []
next_token = None
BATCH_SIZE = 100

while True:
    print(f"Mengambil review ke-{len(all_reviews) + 1} sampai {len(all_reviews) + BATCH_SIZE}...")

    result, next_token = reviews(
        package_name,
        lang='id',
        country='id',
        sort=Sort.NEWEST,
        count=BATCH_SIZE,
        continuation_token=next_token
    )

    if not result:
        print("Tidak ada review lagi.")
        break

    for r in result:
        review_date = r['at']

        # Jika review lebih tua dari start_date, hentikan pengambilan
        if review_date < start_date:
            print("Mencapai batas tanggal awal, berhenti...")
            next_token = None
            break

        # Jika review ada di antara rentang tanggal yang diminta
        if start_date <= review_date <= end_date:
            all_reviews.append(r)

    if not next_token:
        break

    time.sleep(1)

# Konversi ke DataFrame
df = pd.DataFrame([{
    "ReviewID": r.get("reviewId"),
    "Username": r.get("userName"),
    "UserImage": r.get("userImage"),
    "Content": r.get("content"),
    "Score": r.get("score"),
    "ThumbsUpCount": r.get("thumbsUpCount"),
    "ReviewCreatedVersion": r.get("reviewCreatedVersion"),
    "at": r.get("at"),
    "ReplyContent": r.get("replyContent"),
    "RepliedAt": r.get("repliedAt"),
    "AppVersion": r.get("reviewCreatedVersion"),
} for r in all_reviews])

# Simpan ke CSV
output_path = f'../data/review_genshin_{version}.csv'
df.to_csv(output_path, index=False, encoding='utf-8-sig')

print(f"Selesai! {len(df)} review disimpan antara {start_date.strftime('%Y-%m-%d')} dan {end_date.strftime('%Y-%m-%d')}.")
```
